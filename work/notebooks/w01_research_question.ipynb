## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Decision:** Given limited content-team bandwidth, which pages should be
reviewed first for refresh, expansion, or monitoring this month?

**Who acts on it:** A content or SEO team lead deciding where to allocate
their team's next sprint.

**Action:** The output is a ranked queue of candidate pages with reason codes
(e.g. stale-but-visible, declining-with-demand, thin-content) — a human
reviewer decides what to actually do with each one; the model doesn't act
automatically.

**Cost of a wrong call:** A false positive wastes reviewer hours refreshing
content that wasn't actually a problem. A false negative means a genuinely
declining, high-traffic page gets skipped and keeps losing visibility
unaddressed — the more expensive failure mode, since traffic loss compounds
over time.
